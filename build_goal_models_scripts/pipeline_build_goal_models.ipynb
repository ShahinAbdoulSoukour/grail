### Leaf goals classification into TECH, PHYS, and ORG

```python
files = glob.glob("goal_models_NIST/*.json") # <-- change the path

# Exclude files ending with these suffixes
exclude_suffixes = ("_leaf_goals_classified.json", ".operationalized_embeddings.json", ".operationalized_qwen.json")
files = [p for p in files if not p.endswith(exclude_suffixes)]

for path in files:
    with open(path, encoding="utf-8") as f:
        model = json.load(f)

    # classify the leaf goals in the goal model
    classify_leaf_goal(model)

    # print leaf goals
    leaves = leaf_goals_of(model)
    for leaf in leaves:
        print(f"{leaf['goal_id']} [{leaf['class']}] {leaf['goal_name']}")

    n_org = sum(1 for lf in leaves if lf["class"] == "ORG")
    print(f"\n{len(leaves)} leaf goals: {n_org} ORG, {len(leaves) - n_org} TECH")

    out = path.rsplit(".json", 1)[0] + "_leaf_goals_classified.json"
    with open(out, "w", encoding="utf-8") as f:
        json.dump(model, f, indent=2, ensure_ascii=False)
    print(f"--> {out}")
```

### Tests - Traceability

```python
query = """
PREFIX gm: <https://example.org/goal-model/ontology/>

SELECT
    ?standard
    ?goalId
    ?goalName
    ?operationId
    ?operationName
    ?source
    ?score
WHERE {
    ?goal a gm:LeafGoal ;
          gm:standardName ?standard ;
          gm:goalId ?goalId ;
          gm:goalName ?goalName ;
          gm:operationalizedBy ?operation .

    ?operation gm:operationId ?operationId ;
               gm:operationName ?operationName ;
               gm:source ?source .

    << ?goal gm:operationalizedBy ?operation >>
        gm:similarityScore ?score .
}
ORDER BY ?standard ?goalId DESC(?score)
"""
```

```python
import pandas as pd

results = store.query(query)

rows = [
    {
        "Standard": row["standard"].value,
        "Goal ID": row["goalId"].value,
        "Leaf Goal": row["goalName"].value,
        "Operation ID": row["operationId"].value,
        "Operation": row["operationName"].value,
        "Source": row["source"].value,
        "Score": float(row["score"].value),
    }
    for row in results
]

df = pd.DataFrame(rows)

df
```

#### Operation → Root Goal

```python
query = """
PREFIX gm: <https://example.org/goal-model/ontology/>

SELECT
    ?source
    ?operationId
    ?operationName
    ?score
    ?standard
    ?leafGoalId
    ?leafGoalName
    ?objectiveGoalId
    ?objectiveGoalName
    ?rootGoalId
    ?rootGoalName
WHERE {
    # Operation
    ?operation a gm:Operation ;
               gm:source ?source ;
               gm:operationId ?operationId ;
               gm:operationName ?operationName .

    # Leaf Goal operationalized by the operation
    ?leafGoal a gm:LeafGoal ;
              gm:standardName ?standard ;
              gm:goalId ?leafGoalId ;
              gm:goalName ?leafGoalName ;
              gm:parentGoal ?objectiveGoal ;
              gm:operationalizedBy ?operation .

    # Score attached to the Goal -> Operation relation
    << ?leafGoal gm:operationalizedBy ?operation >>
        gm:similarityScore ?score .

    # Objective Goal
    ?objectiveGoal a gm:ObjectiveGoal ;
                   gm:goalId ?objectiveGoalId ;
                   gm:goalName ?objectiveGoalName ;
                   gm:parentGoal ?rootGoal .

    # Root Goal
    ?rootGoal a gm:RootGoal ;
              gm:goalId ?rootGoalId ;
              gm:goalName ?rootGoalName .
}
ORDER BY ?operationId ?standard ?rootGoalId ?leafGoalId DESC(?score)
"""
```

```python
results = store.query(query)

rows = [
    {
        "Source": row["source"].value,
        "Operation ID": row["operationId"].value,
        "Operation": row["operationName"].value,
        "Score": float(row["score"].value),
        "Standard": row["standard"].value,
        "Leaf Goal ID": row["leafGoalId"].value,
        "Leaf Goal": row["leafGoalName"].value,
        "Objective Goal ID": row["objectiveGoalId"].value,
        "Objective Goal": row["objectiveGoalName"].value,
        "Root Goal ID": row["rootGoalId"].value,
        "Root Goal": row["rootGoalName"].value,
    }
    for row in results
]

df = pd.DataFrame(rows)

df
```
